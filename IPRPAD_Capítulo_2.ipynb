###**Sección 2.2.4 Date**

```python
#importación de librería
from datetime import datetime
```

```python
#obtener la fecha actual
datetime.now()
```

```python
#se define una fecha como string (podría ser un tipo object y sería lo mismo)
FECHA='2026-6-26 20:08:07' #siempre observar el orden de año, mes, día para el idioma inglés.
FECHA
```

```python
#convertir a DateTime: internamente es un número (como sucede en excel) y por tanto se podrá hacer operaciones aritméticas
DATETIME = datetime.strptime(FECHA, '%Y-%m-%d %H:%M:%S') # FECHA podría ser un tipo object y sería lo mismo

print(DATETIME) #fecha completa
print(DATETIME.date()) #Año, mes, dia
print(DATETIME.year) #Año
print(DATETIME.month) #mes
print(DATETIME.day) #dia
print(DATETIME.weekday()) #número de día de la semana
print(DATETIME.hour) #hora
print(DATETIME.minute) #minuto
print(DATETIME.second) #segundo
```

```python
#Si es DateTime se puede convertir a string y obtener otras variantes
fecha=DATETIME.strftime('%Y-%m-%d %H:%M:%S')
print(fecha)

print(DATETIME.strftime('%y')) #número año acotado
print(DATETIME.strftime('%Y')) #número año completo

print(DATETIME.strftime('%m')) #número mes
print(DATETIME.strftime('%b')) #nombre mes acotado
print(DATETIME.strftime('%B')) #nombre mes completo

print(DATETIME.strftime('%a')) #nombre del día acotado
print(DATETIME.strftime('%A')) #nombre del día completo
print(DATETIME.strftime('%j')) #número del día en el año

print(DATETIME.strftime('%W')) #número de semana del año

print(DATETIME.strftime('%H')) #hora
print(DATETIME.strftime('%M')) #minuto
print(DATETIME.strftime('%S')) #segundo
```

**Ejemplo 2.3.4.2**

Cree un programa donde a partir de la cantidad producida [500, 300, 200] celulares, tablets y computadoras respectivamente; los insumos requeridos para la producción de cada producto por unidad (chip, pantalla, horas de trabajo respectivamente):
calulares: [1,1,2]
tablets: [1,1,3]
camputadoras:[2,1,5]
y los costos monetarios de producción [45,30,15]
devuelva el costo total de producción involucrado


```python
# Se definen los vectores
P = np.array([500, 300, 200])
V_celular = np.array([1, 1, 2])
V_tablet   = np.array([1, 1, 3])
V_computadora   = np.array([2, 1, 5])
Costo_prod = np.array([45, 30, 15])

# Se calcula el total de recursos necesarios
total_recursos = (P[0] * V_celular) + (P[1] * V_tablet) + (P[2] * V_computadora)

# Se calcula el costo total de producción
costo_total = np.dot(total_recursos, Costo_prod)

print('El costo total de producción es $'+str(costo_total))
```

**Ejemplo 2.3.4.2**

Un inversor posee \$12.000 para invertir en tres fondos diferentes. Los fondos presentan los siguientes rendimientos: fondo de renta fija paga un 3 % de interés anual; fondo de renta variable paga un 4 % anual; fondo de criptoactivos que pagan un 7 % anual. Sabiendo que invertirá \$4.000 más en criptoactivos que en fondos variables y que el interés total a obtener en un año es $670. ¿Qué monto invertirá en cada tipo de fondo?

```python
A=np.array([[1,1,1],
            [0,-1,1],
            [0.03,0.04,0.07]])

invA=np.linalg.inv(A)
d=np.array([12000,4000,670])
x=np.dot(invA,d)
print('La distribución de los montos a invertir son: $'+str(round(x[0]))+' en fondos de renta fija, $'+str(round(x[1]))+' en fondos de renta variable, $'+str(round(x[2]))+' en fondos de criptoactivos')
```
