```python
import numpy as np
matriz=np.array([[0.3, 0.2, 0.1, 0.4, 0.0], [0.3,0.3,0.1,0.2,0.1], [0,0,0.4,0.5,0.1],[0,0,0,1,0],[0,0,0,0,1]])
matriz_N=matriz[0:3,0:3]
matriz_I=np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]])
inv=matriz_I-matriz_N
inversa=np.linalg.inv(inv)
matriz_A=matriz[0:3,3:6]
matriz_PA=inversa @ matriz_A
vector6=np.array([[150000,60000,25000]])
vector_esp=np.dot(vector6,matriz_PA)
vector_esp=np.round(vector_esp,0)
resultado='Se espera que $'+ str(vector_esp[0][0]) + ' sean pagados y que $' + str(vector_esp[0][1]) + ' sean incobrables'
resultado
```

```python
def proveedor(email):
  '''
  Función que devuelve el proveedor de un email
  Paramétros:
    email: un string con el email
  Devuelve: El proveedor del email
  '''
  sep=email.split('@')
  sep2=sep[1].split('.')
  resultado=[]
  if sep2[0].upper()=='GMAIL':
    resultado.append('GOOGLE')
  elif sep2[0].upper()=='OUTLOOK':
    resultado.append('MICROSOFT')
  elif sep2[0].upper()=='HOTMAIL':
    resultado.append('MICROSOFT')
  elif sep2[0].upper()=='YAHOO':
    resultado.append('YAHOO')
  else:
    resultado.append('NO IDENTIFICADO')
  return resultado
```

```python
def imputar_cero(lista):
  '''
  Función que imputa ceros a los valores vacíos de una lista
  Paramétros:
    lista: una secuencia de números
    Devuelve: una lista con los valores vacíos completos con cero
  '''
  lista=lista.split(',')
  lista=[int(i) if i.isdigit() else None for i in lista]
  for i in range(len(lista)):
    if lista[i] == None:
      lista[i]= 0
    else:
      pass
  return lista
```

**Ejercicio 3.5.3**

```python
def pedido(prod, cant):
  '''
  Función que devuelve el detalle del pedido
  Paramétros:
    prod: un string con los productos
  Devuelve: El detalle del pedido y el total a pagar
  '''
  menu=dict({'1': ['Empanada de carne',500],'2': ['Empanada de jamón y queso',500],
                '3': ['Empanada de humita',500],'4': ['Pizza de muzzarela',15000],
                '5': ['Pizza de napolitana',18000],'6': ['Pizza de fugazetta',17000]})
  producto=prod.split(',')
  cantidades=cant.split(',')
  cantidades=[int(i) for i in cantidades]
  prod_cant=dict(zip(producto,cantidades))
  pedido=[]
  for i in menu:
    for j in prod_cant:
      if i==j:
        pedido.append(str(prod_cant[i])+' '+ str(menu[i][0]+' '+str(menu[i][1])))
  total=[]
  for i in prod_cant:
    for j in menu:
      if i==j:
        total.append(prod_cant[i]*menu[j][1])
  gasto_total=sum(total)
  pedido.append('El total a pagar es '+str(gasto_total))
  return pedido
```

```python
pedido(input('Ingrese códigos de productos: '), input('Ingrese cantidades: '))
```

**Ejercicio 3.5.4**

```python
def indices(cant_ant, precio_ant, cant_act, precio_act):
  '''
  Función que calcula Índices agregados ponderados
  Paramétros:
    cant_ant: cantidades mes anterior
    precio_ant: precios del mes anterior
    cant_act: cantidades mes actual
    precio_act: cantidades del mes actual
  Devuelve: los índices de Laspeyre, Paschee y Fisher
  '''
  cant_ant=cant_ant.split(',')
  cant_ant=[float(i) for i in cant_ant]
  precio_ant=precio_ant.split(',')
  precio_ant=[float(i) for i in precio_ant]
  cant_act=cant_act.split(',')
  cant_act=[float(i) for i in cant_act]
  precio_act=precio_act.split(',')
  precio_act=[float(i) for i in precio_act]
  numeradorL=[]
  for i,j in zip(cant_ant,precio_act):
    numeradorL.append(i*j)
  denominadorL=[]
  for i,j in zip(cant_ant,precio_ant):
    denominadorL.append(i*j)
  IL=round((sum(numeradorL)/sum(denominadorL))*100,2)
  numeradorP=[]
  for i,j in zip(cant_act,precio_act):
    numeradorP.append(i*j)
  denominadorP=[]
  for i,j in zip(cant_act,precio_ant):
    denominadorP.append(i*j)
  IP=round((sum(numeradorP)/sum(denominadorP))*100,2)
  IF=round((IL*IP)**(1/2),2)
  resultado='El IL es '+str(IL)+', el IP es '+str(IP)+' y el IF es '+str(IF)
  return resultado
```

```python
indices(input('Ingrese cantidades consumidas en el mes anterior: '),
        input('Ingrese precios del mes anterior: '),
        input('Ingrese cantidades consumidas en el mes actual: '),
        input('Ingrese precios del mes actual: '))
```
