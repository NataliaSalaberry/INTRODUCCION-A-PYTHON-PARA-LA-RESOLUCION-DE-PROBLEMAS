## **Sección 3.1 Condicional IF**

```python
#definimos una variable sobre la cual operar
x = True

if x==True:
  print('x es verdadero!')
```

```python
x = False

if x==True:
  print('x es verdadero!')
```

Como puedo observarse al ejectar, si no se cumple la condición no devuelve nada. Entonces, es posible completar la alternativa del siguiente modo

```python
x = False

if x==True:
  print('x es verdadero!')
else:
  print('x no es verdadero!')
```

También es posible agregar más niveles de condiciones a evaluar utilizando elif

```python
x = 'Banco'

if x == 'Veterinaria':
    print('Bienvenido a la Veterinaria!')
elif x == 'Banco':
    print('Bienvenido al banco!')
else:
    print('¿Donde estoy?!')
```

También es posible establecer condiciones múltiples a evaluar utilizando operadores

```python
x = 1
y = 2

if x == 1 and y==0:
    print(x+y)
elif x == 1 and y==2:
    print(x+y)
else:
    print('Nada')
```

**Ejemplo 3.1.2**

Cree un programa donde el usuario ingrese un número y devuelva si es número primo o no.

Recordar:  
*   Si es un número negativo, 0 o 1, no es primo
*   Si es 2 es primo
*   Si es mayor que 2 y divisible por 3, 5 o 7, es primo
*   Si es mayor que 2 y múltiplo de 2, 3, 5 o 7, no es primo
*   Si es mayor que 2 y “en cualquier otro caso”, es primo
*   Solo los números enteros positivos pueden ser primo.




```python
num=input('Introduzca un número: ')
if num.isdigit():
  x=int(num)
  if x < 0 or x==0 or x==1: #si negativo, cero o 1 no es primo
    print('El número '+str(x)+' no es primo')
  elif x==2: #si es 2 es primo
    print('El número '+str(x)+' es primo')
  elif x>2 and x%x==0 and (x/3==1 or x/5==1 or x/7==1): #si es mayor a 2 y es divisible por 3 o 5 o 7, es primo
    print('El número '+str(x)+' es primo')
  elif x>2 and x%x==0 and (x % 2==0 or x % 3==0 or x % 5==0 or x % 7==0) : #si es mayor a 2 y es múltiplo de 2 o de 3 o de 5 o de 7, no es primo
    print('El número '+str(x)+' no es primo')
  else:
    print('El número '+str(x)+' es primo')
else:
  print('Debe ingresar un número entero positivo')
```

```python
#resolución del ejercicio 1.1.2
def bisiesto(x):
  '''
      Función que determina si un año es bisiesto
      Parámetros:
        año: número entero correspondiente a un año
      Devuelve:
        bisiesto o no bisiesto.
  '''
  if not isinstance(x, int):
        print(f"{x} no es un año válido.")
        return


  if x is int and x>=1:
    resultado="es bisiesto"
  elif x%400==0:
    resultado="es bisiesto"
  elif x%4==0 and x%100!=0:
    resultado="es bisiesto"
  else:
    resultado="no es bisiesto"
  return resultado
```

```python
#se aplica la función creada
bisiesto(2026)
```

```python
#se aplica la función creada
bisiesto(2024)
```

```python
#se aplica la función creada
bisiesto(2026.5)
```

```python
#también es posible aplicarla en la creación de un programa

año=int(input('Ingrese un año: '))
print(bisiesto(año))
```

```python
def mayor(x,y):
  '''
      Función que determina si un número es mayor que otro
      Parámetros:
        x: primer número
        y: segundo número
      Devuelve:
        cual número es mayor o que se ingresen números distintos si se ingresó números iguales.
  '''
  if x!=y:
    if x>y:
      resultado= "x="+str(x)+" es mayor que y="+str(y)
    else:
      resultado="y="+str(y)+" es mayor que x="+str(x)
  else:
    resultado="debe ingresar dos números distintos"
  return resultado
```

```python
#se aplica la función creada
mayor(3,2)
```

```python
#se aplica la función creada
mayor(3,3)
```

```python
#se aplica la función creada
mayor(2,3)
```

```python
def datos_personales(nombre, apellido, nacionalidad, edad):
  '''
      Función que devuelve datos personales
      Parámetros:
        nombre: nombre ingresado por el usuario
        apellido: apellido ingresado por el usuario
        nacionalidad: nacionalidad ingresada por el usuario
        edad: edad ingresada por el usuario
      Devuelve:
        El ciudadano (nombre y apellido) de nacionalidad (nacionalidad) nació en el año (año nacimiento).
  '''

  nombre=nombre.upper()
  apellido=apellido.upper()
  nacionalidad=nacionalidad.upper()
  import datetime
  hoy=datetime.date.today()
  año_actual=hoy.year
  año_nacimiento=año_actual-int(edad)
  resultado='El ciudadano '+ str(nombre) + ' ' + str(apellido) + ' de nacionalidad ' + str(nacionalidad) + ' nació en el año '+str(año_nacimiento)
  return resultado
```

```python
def markov_esperado(vector1, vector2, vector3, vector4, vector5, vector6):
  '''
      Función que devuelve el monto total esperado que sea pagado y el monto total que será incobrable
      Parámetros:
        vector1: primera fila de la matriz de transición
        vector2: segunda fila de la matriz de transición
        vector3: tercera fila de la matriz de transición
        vector4: cuarta fila de la matriz de transición
        vector5: quinta fila de la matriz de transición
        vector6: vector inicial
      Devuelve:
        Se espera que (MONTO PAGADOS) sean pagados y que (MONTO INCOBRABLE) sean incobrables.
  '''

  import numpy as np
  matriz=np.array([vector1, vector2, vector3,vector4,vector5])
  matriz_N=matriz[0:3,0:3]
  matriz_I=np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]])
  inv=matriz_I-matriz_N
  inversa=np.linalg.inv(inv)
  matriz_A=matriz[0:3,3:6]
  matriz_PA=inversa @ matriz_A
  vector6=np.array([vector6])
  vector_esp=np.dot(vector6,matriz_PA)
  vector_esp=np.round(vector_esp,0)
  resultado='Se espera que $'+ str(vector_esp[0][0]) + ' sean pagados y que $' + str(vector_esp[0][1]) + ' sean incobrables'
  return resultado
```

```python
markov_esperado([0.3, 0.2, 0.1, 0.4, 0.0],
                 [0.3,0.3,0.1,0.2,0.1],
                 [0,0,0.4,0.5,0.1],
                 [0,0,0,1,0],
                 [0,0,0,0,1],
                 [150000,60000,25000])
```
